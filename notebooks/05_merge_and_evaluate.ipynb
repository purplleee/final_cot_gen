# 05 Merge and Evaluate

Merge clean CoT outputs from contiguous Groq/Ollama shards, join prompt metadata, parse `Verdict:`, and write a generation report. No poisoning, LLM-Merge, ASR, or training is done here.


This cell imports merge/evaluation dependencies and defines the final output paths for merged JSONL and the generation report CSV.

```python
from pathlib import Path
import json
import re

import pandas as pd
from IPython.display import display

PROJECT_ROOT = Path("..").resolve() if Path.cwd().name == "notebooks" else Path(".").resolve()
# CHANGED: generated/intermediate outputs stay on E: to avoid filling the C: drive.
OUTPUT_ROOT = Path(r"E:\PFE_CoT_outputs")
INTERIM_DIR = OUTPUT_ROOT / "data" / "interim"
FINAL_DIR = OUTPUT_ROOT / "data" / "final"
MERGED_OUTPUT_PATH = FINAL_DIR / "merged_clean_cot_outputs.jsonl"
REPORT_PATH = FINAL_DIR / "clean_cot_generation_report.csv"

FINAL_DIR.mkdir(parents=True, exist_ok=True)

print("Prompt folder:", INTERIM_DIR)
print("Final folder:", FINAL_DIR)
print("Merged output:", MERGED_OUTPUT_PATH)
print("Report output:", REPORT_PATH)
```

This cell defines JSONL read/write helpers plus verdict parsing and label normalization utilities used during evaluation.

```python
def load_jsonl(path):
    records = []
    if not path.exists():
        return records
    with open(path, "r", encoding="utf-8") as input_handle:
        for line in input_handle:
            line = line.strip()
            if line:
                records.append(json.loads(line))
    return records


def write_jsonl(path, records):
    with open(path, "w", encoding="utf-8") as output_handle:
        for record in records:
            output_handle.write(json.dumps(record, ensure_ascii=False) + "\n")


def parse_verdict(text):
    if not isinstance(text, str) or text == "failed_generation":
        return "failed_generation" if text == "failed_generation" else "unknown"
    match = re.search(r"(?im)^\s*Verdict\s*:\s*(?:\n\s*)?([^\r\n]+)", text)
    if match:
        return match.group(1).strip().strip("`*")
    return "unknown"


def normalize_label(value):
    if value is None or pd.isna(value):
        return "unknown"
    text = str(value).strip().lower()
    text = re.sub(r"\s+", " ", text)
    return text.strip(" .;:,_`*")
```

This cell loads all shard prompt JSONL files and indexes prompt metadata by stable `original_id`.

```python
prompt_files = sorted(INTERIM_DIR.glob("reasoning_prompts_shard*.jsonl"))
if not prompt_files:
    raise FileNotFoundError("No shard prompt JSONL files found. Run notebook 02 for each shard you want to merge.")

prompt_records = []
for path in prompt_files:
    records = load_jsonl(path)
    prompt_records.extend(records)
    print(f"Loaded {len(records)} prompt records from {path.name}")

missing_original_ids = [record for record in prompt_records if not record.get("original_id")]
if missing_original_ids:
    raise KeyError("Prompt records must contain original_id. Re-run notebook 02 after updating the pipeline.")

prompt_by_original_id = {str(record["original_id"]): record for record in prompt_records}
all_prompt_original_ids = set(prompt_by_original_id)
all_prompt_sample_ids = {int(record["sample_id"]) for record in prompt_records if "sample_id" in record}

print("Unique prompt original_ids:", len(all_prompt_original_ids))
print("Unique temporary sample_ids:", len(all_prompt_sample_ids))
if len(prompt_by_original_id) != len(prompt_records):
    print("WARNING: duplicate original_id values found in prompt records; final merge keeps the last prompt metadata for duplicates.")
```

This cell discovers Groq and Ollama final output files and loads all generated records for merging.

```python
output_files = sorted(FINAL_DIR.glob("groq_*_shard*.jsonl")) + sorted(FINAL_DIR.glob("ollama_*_shard*.jsonl"))
output_files = [path for path in output_files if "checkpoint" not in path.name and path.name != MERGED_OUTPUT_PATH.name]

print("Output files found:")
for path in output_files:
    print("-", path.name)

all_records = []
for path in output_files:
    for record in load_jsonl(path):
        record["source_output_file"] = path.name
        all_records.append(record)

print("Generated records loaded:", len(all_records))
```

This cell merges generation records by backend, model, and `original_id`, then joins true labels and metadata from the prompt records.

```python
# Keep the last record for each backend/model/original_id, but report duplicates first.
if all_records:
    raw_df = pd.DataFrame(all_records)
    if "original_id" not in raw_df.columns:
        raise KeyError("Generation outputs must contain original_id. Re-run notebooks 03/04 after updating the pipeline.")
    raw_df["sample_id"] = raw_df["sample_id"].astype(int)
    raw_df["original_id"] = raw_df["original_id"].astype(str)
    duplicate_mask = raw_df.duplicated(subset=["backend", "model", "original_id"], keep=False)
    duplicate_count = int(duplicate_mask.sum())
    print("Duplicate backend/model/original_id records:", duplicate_count)
    if duplicate_count:
        display(raw_df.loc[duplicate_mask, ["backend", "model", "original_id", "sample_id", "source_output_file"]].sort_values(["backend", "model", "original_id"]).head(30))

merged_by_key = {}
for record in all_records:
    if not record.get("original_id"):
        raise KeyError(f"Missing original_id in generated record with sample_id={record.get('sample_id')}")
    key = (record.get("backend", "unknown"), record.get("model", "unknown"), str(record["original_id"]))
    merged_by_key[key] = record

merged_records = []
for record in merged_by_key.values():
    original_id = str(record["original_id"])
    sample_id = int(record["sample_id"])
    prompt_record = prompt_by_original_id.get(original_id, {})
    generation_text = record.get("raw_generation") or record.get("reasoning", "")
    parsed_verdict = parse_verdict(generation_text)
    if parsed_verdict == "unknown" and record.get("prediction"):
        parsed_verdict = record.get("prediction")

    enriched = dict(record)
    enriched.update({
        "sample_id": sample_id,
        "original_id": original_id,
        "true_label": prompt_record.get("true_label", "unknown"),
        "category": prompt_record.get("category", "unknown"),
        "family": prompt_record.get("family", "unknown"),
        "parsed_verdict": parsed_verdict,
    })
    true_norm = normalize_label(enriched["true_label"])
    verdict_norm = normalize_label(enriched["parsed_verdict"])
    enriched["agreement_with_true_label"] = (
        verdict_norm == true_norm
        if true_norm != "unknown" and verdict_norm not in {"unknown", "failed_generation"}
        else None
    )
    merged_records.append(enriched)

merged_records = sorted(merged_records, key=lambda record: (record.get("backend", ""), record.get("model", ""), str(record.get("original_id", ""))))
write_jsonl(MERGED_OUTPUT_PATH, merged_records)
print("Merged records written:", len(merged_records))
print("Saved:", MERGED_OUTPUT_PATH)
```

This cell reports missing samples, failed generations, genuine parsed-verdict agreement, and saves the clean CoT generation report.

```python
report_rows = []
if merged_records:
    merged_df = pd.DataFrame(merged_records)
    merged_df["sample_id"] = merged_df["sample_id"].astype(int)
    merged_df["original_id"] = merged_df["original_id"].astype(str)
    merged_df["is_failed_generation"] = merged_df["reasoning"].eq("failed_generation") | merged_df["parsed_verdict"].eq("failed_generation")
    merged_df["has_parsed_verdict"] = ~merged_df["parsed_verdict"].isin(["unknown", "failed_generation", ""])

    generated_any_ids = set(merged_df["original_id"])
    missing_any_ids = sorted(all_prompt_original_ids - generated_any_ids)

    print("Missing original_ids across all outputs:", len(missing_any_ids))
    print("Failed generations:", int(merged_df["is_failed_generation"].sum()))
    print("Records with genuine parsed model verdict:", int(merged_df["has_parsed_verdict"].sum()))

    if missing_any_ids:
        print("First missing original_ids:", missing_any_ids[:30])

    display_columns = ["sample_id", "original_id", "backend", "model", "true_label", "category", "family", "parsed_verdict", "agreement_with_true_label"]
    display(merged_df[display_columns].head())

    for (backend, model), group in merged_df.groupby(["backend", "model"], dropna=False):
        generated_ids = set(group["original_id"])
        missing_ids = sorted(all_prompt_original_ids - generated_ids)
        comparable = group[group["agreement_with_true_label"].notna()]
        agreement_rate = comparable["agreement_with_true_label"].mean() if len(comparable) else None
        report_rows.append({
            "backend": backend,
            "model": model,
            "total_records": len(group),
            "unique_original_ids": group["original_id"].nunique(),
            "unique_sample_ids_debug_only": group["sample_id"].nunique(),
            "duplicate_records_after_merge": len(group) - group["original_id"].nunique(),
            "missing_original_ids_vs_prompt_set": len(missing_ids),
            "failed_generations": int(group["is_failed_generation"].sum()),
            "genuine_parsed_verdict_records": int(group["has_parsed_verdict"].sum()),
            "agreement_count": int(group["agreement_with_true_label"].eq(True).sum()),
            "agreement_rate_when_comparable": agreement_rate,
            "output_files": "; ".join(sorted(group["source_output_file"].dropna().unique())),
        })

        print(f"\n{backend} | {model}")
        print("Records:", len(group))
        print("Missing original_ids vs all prompt shards:", len(missing_ids))
        print("Failed generations:", int(group["is_failed_generation"].sum()))
        if len(comparable):
            print("Agreement with true_label using genuine parsed model verdict:", round(float(agreement_rate), 4))
            display(pd.crosstab(comparable["true_label"], comparable["parsed_verdict"], dropna=False))
        else:
            print("Agreement with true_label using genuine parsed model verdict: not enough comparable verdicts")
else:
    print("No generation outputs found yet. Run notebook 03 and/or notebook 04 first.")

report_df = pd.DataFrame(report_rows)
report_df.to_csv(REPORT_PATH, index=False)
print("Report saved:", REPORT_PATH)
if not report_df.empty:
    display(report_df)
```
